```python
columns = ["Rating","gross margin percentage", "gross income","Unit price","Tax 5%","Total","Payment","cogs"]
df[columns] = df[columns].apply(lambda x: pd.to_numeric(x.astype(str).str.replace(',', '.'), errors='coerce'))
```

```python
df.dtypes
```

```python
df["Customer type"] = df["Customer type"].astype(str)
```

```python
df["Gender"] = df["Gender"].astype(str)
```
